## Lesson

### What DMAIC is

DMAIC is the project structure at the core of Six Sigma, a quality-improvement method started at Motorola in 1986 and made famous by GE in the 1990s. It is meant for **improving an existing process** whose performance is measurable but whose cause of trouble is unknown. (Its sister, DMADV (Define, Measure, Analyze, Design, Verify), is for designing a new process.)

The five phases always run in this order. Each one answers one question. (The techniques in the last column are each defined in their own module.)

| Phase | The question it answers | It must deliver (the tollgate list) | Techniques in this playbook |
|---|---|---|---|
| **Define** | What exactly is the problem, and is it worth a project? | problem statement, scope (in / out), goal, SIPOC process map | Pareto analysis (Module 2) |
| **Measure** | Can we trust the numbers, and what is normal? | operational definitions, a check of the data itself, baseline with its normal variation | timeline reconstruction (Module 3), is / is-not (Module 4) |
| **Analyze** | Why does it happen? | located problem, candidate causes, each verified or rejected with data; mechanism vs. root cause | change analysis (Module 4), 5 Whys (5), fishbone (6), cause mapping (7), fault tree (8) |
| **Improve** | What fixes it, and did the fix work? | chosen fix, tested, improvement shown in the Define measure | barrier analysis (Module 9) |
| **Control** | How do we keep it fixed? | control plan: monitor, limit, owner, response; handover | FMEA, Failure Mode and Effects Analysis (Module 10) |

### Define

- **Problem statement:** what, where, when, how big. It never contains *why* (that is Analyze) or *the fix* (that is Improve). "The share of starters who became closed-won deals fell from 0.087 to 0.058 on 18–19 Oct" is a problem statement. "Closed-won fell because of a campaign change" is a guess dressed as one; "closed-won fell because the lead-form step broke" already contains an Analyze result.
- **Scope:** what is in and out of the project, so it does not grow forever.
- **Goal:** the measurable target that means "solved".
- **SIPOC (Suppliers, Inputs, Process, Outputs, Customers):** a one-line map of the process: who feeds it, what goes in, the main steps, what comes out, and who receives it. It fixes the boundary of what is being studied.

### Measure

- **Operational definition:** exactly how a number is counted (which rows, which date, what counts as "reached").
- **Check the measurement before trusting it:** does the data reproduce an independent total, are there impossible values? In manufacturing this step is called measurement system analysis; for an analyst, it means checking the data itself.
- **Baseline with its normal variation:** a rate is never perfectly flat. The baseline must say how much it *normally* moves, so a real change can be told apart from noise. The p-chart below does exactly that.

### The p-chart, step by step

A p-chart (a control chart for a proportion, from statistical process control) asks: *is today's rate further from normal than chance would explain?*

1. Pick a baseline period when the process was normal. Its overall rate is the centre line, **p̄**.
2. Each day has its own number of trials **n** (e.g. users who started onboarding).
3. Chance alone moves a day's rate by about one **standard error**: √(p̄ × (1 − p̄) / n). So each day gets its own standard error: a day with more starters gets a smaller one (tighter limits), a day with fewer gets a larger one (wider limits).
4. The control limits are p̄ ± 3 standard errors for that day.
5. A day outside the limits is a **signal**: the process probably changed. A day inside is treated as normal variation.

Why 3? It is the standard convention (from Walter Shewhart, who invented control charts in the 1920s). If the day's rate is close to normally distributed (true when n is large, as here), a normal day falls outside ±3 standard errors by chance only about 0.27% of the time (about 1 day in 370), so false alarms are rare. This is a convention, not a law. Tighter limits catch smaller changes but raise more false alarms.

### Analyze

- **Locate** the problem: which step, which segment, which days.
- **List candidate causes**, then **verify or reject each with data**. A cause nobody tested is still a Hypothesis.
- **Mechanism vs. root cause.** The mechanism is *how* the number moved ("the traffic mix shifted toward a low-converting group"). The root cause is *why, in the real world* ("someone edited the campaign's device targeting"). Data often proves the mechanism and can only point at the cause.

### Improve

- Choose a fix aimed at the verified cause, test it (on a small scale first if possible), and show the improvement **in the same measure that defined the problem**. A fix judged on a different metric has not been shown to fix this problem.

### Control

- A **control plan** lists, for each thing to watch: the measure, the limit that triggers action, the owner, and the response. Without it, a fixed problem can come back unnoticed.
- The monitor should watch the **cause**, not only the headline number. A headline alarm says "something is wrong"; a cause-level alarm says "this specific thing is wrong again".

### Tollgates

At the end of each phase, a **tollgate review** checks the phase's deliverables (the table above) before the next phase starts. Its purpose is to stop a project from, for example, analysing causes of a problem that the Measure phase would have shown to be a logging error.

### When DMAIC fits, and when it does not

| Situation | DMAIC? | Why |
|---|---|---|
| An existing, measurable process performs worse than it should, cause unknown | **Yes** | This is exactly its purpose |
| A problem is hurting customers *right now* and must be contained today | **Not first** | DMAIC has no containment step; 8D (Eight Disciplines, Module 11) does |
| The cause is already known and the fix is obvious | **No** | Just fix it; a five-phase project is overhead |
| A new process is being designed | **No** | Use DMADV |

## Worked example: the E01 funnel drop as a DMAIC project

**Data choice: reuse.** E01 was a funnel-drop investigation that covered most of DMAIC without naming it. Re-running it phase by phase shows the structure on a real case, and shows where it fell short.

**Where this was already done, unnamed:** the user's analysis workspace, `AI Analyst Workspace/Reports/E01_03_Root_Cause_Analysis.md` and `E01_Summary.md`. These are local files: the workspace has no public repository.

**The data:** a take-home exercise with artificial data: one row per **starter** (a user who started onboarding), with a date for each funnel step reached (blank = not reached), plus campaign, platform and browser. A **lead** is a starter who submitted the lead form. The source file is not in this repository (see the README); `python data/get_data.py` copies it in and checks it.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import common
from common import show, BLUE, ORANGE, AQUA, GREY

d = common.load_e01()
steps = list(common.E01_STEPS)
names = list(common.E01_STEPS.values())
d["cell"] = d["campaign_id"] + " / " + d["Platform"]
d["lead_form"] = d["lead_form_submitted"].notna()
ddmm = lambda day: pd.Timestamp(day).strftime("%d.%m")
print(f"{len(d):,} rows, {d['user_id'].nunique():,} distinct users")
for col in ["campaign_id", "Platform", "browser"]:
    print(f"{col}: " + ", ".join(f"{k} = {v:,}" for k, v in d[col].value_counts().items()))
print(f"onboarding days: {d['ob_day'].min()} to {d['ob_day'].max()}")
print("weekdays:", ", ".join(pd.to_datetime(sorted(d['ob_day'].unique())).day_name()))
```

### Part 0 — Measure, step 1: check the data first

The Measure phase starts with the data itself: the facts that limit what it can say. (This comes before Define here only because the problem statement must quote trusted numbers.)

```python
ref = common.load_e01_reference()
mine = d.groupby("ob_day").agg(users=("user_id", "size"),
                               **{s: (s, lambda x: x.notna().mean()) for s in steps[1:]}).reset_index()
max_gap = np.abs(mine[steps[1:]].to_numpy() - ref[steps[1:]].to_numpy(dtype=float)).max()
same = max_gap < 1e-9 and (mine['users'].to_numpy() == ref['users'].to_numpy()).all()
print(f"Funnel computed from the raw rows ('Main Funnel Data') = the file's ready-made summary ('Funnel table'): "
      f"{'match' if same else 'NO MATCH'}")

skips = sum((d[a].isna() & d[b].notna()).sum() for a, b in zip(steps[1:-1], steps[2:]))
lf_before = (d["lead_form_submitted"] < d["onboarding_start_date"]).sum()
ss_before = (d["strategy_session_call_date"] < d["free_consultation_call_date"]).sum()
day_n = d.groupby("ob_day").size()
print(f"Steps present with the previous step missing: {skips} rows")
print(f"Steps present, but dated before the previous step: {lf_before} rows (lead form before onboarding), "
      f"{ss_before} rows (strategy session before free consultation)")
print(f"Last downstream date: {d[steps[1:]].max().max():%Y-%m-%d}")
for a, b, na, nb in zip(steps[:-1], steps[1:], names[:-1], names[1:]):
    lag = (d[b] - d[a]).dt.days.dropna()
    print(f"  {na} -> {nb}: {(lag == 0).mean():.1%} same day, {(lag > 2).mean():.1%} take more than 2 days, "
          f"longest gap {lag.max():.0f} days ({(lag == lag.max()).sum()} leads)")
print(f"Users per onboarding day: " + ", ".join(f"{ddmm(k)} = {v:,}" for k, v in day_n.items()))
print(f"20.10 has {day_n['2021-10-20'] / day_n.drop('2021-10-20').mean():.0%} of the average of the other five days")
won = d.dropna(subset=["closed_won_date"])
won_lag = (won["closed_won_date"] - won["onboarding_start_date"]).dt.days
print(f"Closed-won deals dated 3 days after onboarding: {(won_lag[won['ob_day'] == '2021-10-15'] == 3).sum()} of "
      f"{(won['ob_day'] == '2021-10-15').sum()} for 15.10 starters, {(won_lag[won['ob_day'] != '2021-10-15'] == 3).sum()} for all other days")
```

- **Observation:** the funnel computed from the raw rows matches the file's own summary sheet, so the counting is trusted.
- **Observation:** no lead skips a step. 9 rows have the lead form dated before onboarding and 28 have the strategy session before the free consultation. These are small date errors; they do not change which steps a lead reached, which is all that the rates below use.
- **Observation:** most leads take each step on the same day as the step before it (84.9% to 100.0%, depending on the step). The longest gap is 7 days, for 30 leads between lead form and free consultation; The data runs to 22 Oct, so starters from 20 Oct had only 2 days to progress. Only a small share of moves take longer than 2 days (printed above), so this cut-off barely affects the rates below.
- **Observation:** almost all of the 15 Oct starters' closed-won deals (94 of 103) are dated 3 days later, on 18 Oct; on every other day, deals are dated the same day or the next. This looks like a dating quirk. It does not affect the rates below, which count each starter on their onboarding day.
- **Observation:** only six onboarding days (Friday 15 Oct to Wednesday 20 Oct). 20 Oct has 61% of the average of the other five days. **Hypothesis:** it is a partial day (data cut off mid-day); a quiet Wednesday is the alternative. Either way it is used for rates, not volumes.
- **Observation:** the data starts at onboarding. Ad impressions, clicks and landing-page views are not in it, so anything that happens before onboarding is invisible.
- **Coverage caveat:** the baseline below rests on three days, one a Friday and two weekend days. The incident days are a Monday and a Tuesday. A weekday effect can't be ruled out from the data alone. The Wednesday recovery (20 Oct) argues against it.

### Part 1 — Define

The daily funnel: the share of each day's starters reaching each step (step-to-step rates follow in Part 3a). The baseline days, 15–17 Oct, are chosen in Part 2.

```python
cnt = d.groupby("ob_day")[steps].count()
daily = pd.DataFrame({"day": [ddmm(x) for x in cnt.index], "started": cnt[steps[0]].to_numpy()})
for s, nm in zip(steps[1:], names[1:]):
    daily[nm] = (cnt[s] / cnt[steps[0]]).map("{:.3f}".format).to_numpy()
show(daily)

base_days = ["2021-10-15", "2021-10-16", "2021-10-17"]
inc_days = ["2021-10-18", "2021-10-19"]
d["won"] = d["closed_won_date"].notna()
B, I = d[d["ob_day"].isin(base_days)], d[d["ob_day"].isin(inc_days)]
w_base, w_inc = B["won"].mean(), I["won"].mean()
print("\nStarted Onboarding → Closed Won rate (closed-won ÷ starters):")
print(f"  15-17 Oct = {B['won'].sum():,} ÷ {len(B):,} = {w_base:.3f}")
print(f"  18-19 Oct = {I['won'].sum():,} ÷ {len(I):,} = {w_inc:.3f}")
print(f"  change = {w_inc / w_base - 1:+.1%}")
```

**Observation:** on 18 and 19 Oct together, the share of starters who became closed-won deals fell from 0.087 to 0.058 (−33.3%), across 3,760 starters. Every column of the table falls on those two days. *Which* step the fall starts at is a question for Analyze (Part 3), not for Define.

**Why the bottom line, not the step where the fall starts:** the original brief reported only "a significant drop in the company's main funnel". Writing "the lead-form step fell" here would put an Analyze result into the problem statement.

**The Define deliverables:**

- **Problem statement:** *the share of starters who became closed-won deals fell from 0.087 (15–17 Oct) to 0.058 on 18–19 Oct (−33.3%), across all 3,760 starters of those two days.* It contains no "why", no fix, and no step.
- **Scope.** In: the six-step funnel from onboarding start to closed-won, 15–20 Oct, all three campaigns. Out: anything before onboarding (not in the data), and the funnel's conversion levels on normal days (for example the low desktop lead-form rates in Part 3b), which are a separate improvement project.
- **Goal:** the closed-won share back inside its normal range (the p-chart limits from Part 2), and a monitor that catches a repeat on its first day.
- **SIPOC**, using only what the data contains:

| Suppliers | Inputs | Process | Outputs | Customers |
|---|---|---|---|---|
| three ad campaigns (AAAA_123, AAAA_456, BBB_123) | visitors, by platform (Desktop / Mobile) and browser (Chrome / Safari) | onboarding → lead form → free consultation → strategy session → sales call → closed-won | lead forms, then closed-won deals | the sales team that runs the calls (the data names it only through its call steps: free consultation, strategy session, sales call) |

### Part 2 — Measure: is the drop real or normal variation?

Part 0 checked the data. Now the rest of Measure: the baseline of the Define measure and its normal variation, with a p-chart. **Chosen baseline:** 15–17 Oct, the three days *before* the drop, because limits must come from a period known to be normal. Three days is a short baseline (caveat above).

```python
def pchart(num, den, base):
    """Per-day rate, centre line from the baseline days, 3-standard-error limits, and z."""
    pbar = num[base].sum() / den[base].sum()
    se = np.sqrt(pbar * (1 - pbar) / den)
    return pd.DataFrame({"n": den, "rate": num / den, "centre": pbar,
                         "lower": pbar - 3 * se, "upper": pbar + 3 * se, "z": (num / den - pbar) / se})


def pchart_table(pc):
    out = pc.copy()
    out.insert(0, "day", [ddmm(x) for x in out.index])
    out["signal"] = np.where(out["z"].abs() > 3, "yes", "no")
    for c in ["rate", "centre", "lower", "upper"]:
        out[c] = out[c].map("{:.4f}".format)
    out["z"] = out["z"].map("{:+.2f}".format)
    show(out)


def pchart_plot(pc, ylabel, title, note, note_xy, ylim):
    fig, ax = plt.subplots(figsize=(7.5, 3.6))
    x = np.arange(len(pc))
    ax.step(x, pc["upper"], where="mid", color=GREY, lw=1, ls="--", label="control limits (±3 standard errors)")
    ax.step(x, pc["lower"], where="mid", color=GREY, lw=1, ls="--")
    ax.axhline(pc["centre"].iloc[0], color=GREY, lw=1, label=f"baseline {pc['centre'].iloc[0]:.3f} (15-17 Oct)")
    sig = pc["z"].abs() > 3
    ax.plot(x, pc["rate"], color=BLUE, lw=1.5)
    ax.scatter(x[~sig], pc["rate"][~sig], color=BLUE, zorder=3, label="day inside the limits")
    ax.scatter(x[sig], pc["rate"][sig], color=ORANGE, zorder=3, s=50, label="day outside the limits")
    ax.set_xticks(x, [ddmm(i) for i in pc.index])
    ax.set_ylabel(ylabel)
    ax.set_title(title, fontsize=10, loc="left")
    ax.text(*note_xy, note, ha="center", fontsize=8.5, color="#333333")
    ax.set_ylim(*ylim)
    ax.legend(fontsize=7.5, loc="upper center", ncol=2, frameon=False)
    plt.tight_layout(); plt.show()


pc_won = pchart(cnt["closed_won_date"], cnt["onboarding_start_date"], base_days)
pchart_table(pc_won)
w_b, st_b = cnt.loc[base_days, "closed_won_date"].sum(), cnt.loc[base_days, "onboarding_start_date"].sum()
print(f"\nCentre line p̄ = closed-won ÷ starters on 15-17 Oct = {w_b:,} ÷ {st_b:,} = {w_b / st_b:.4f}")
```

```python
pchart_plot(pc_won, "share of starters who became closed-won",
            "Closed-won share of starters by onboarding day, p-chart (limits from 15-17 Oct)",
            "18.10 and 19.10 fall below the lower limit;\n20.10 is back at the centre line", (1.0, 0.049), (0.04, 0.135))
```

**How to read the table and chart:** each day's limits differ because each day has a different number of starters. `z` is how many standard errors the day's rate sits from the baseline.

- **Observation:** the three baseline days sit within ±0.81 standard errors of the centre line, as normal days should.
- **Observation:** 18 Oct is 4.61 and 19 Oct is 4.33 standard errors below it. Chance alone puts a normal day outside 3 standard errors about 0.27% of the time, so these two are signals, not noise.
- **Observation:** 20 Oct is back at the centre (+0.08).
- **Conclusion (Measure):** the drop is real, it lasted two days, and the closed-won share was back to normal on 20 Oct (what brought it back is unknown).

### Part 3 — Analyze

**3a. Locate the step.** Each step's rate as a share of the *previous* step: by day, then for the baseline and incident days pooled.

```python
step_rate = pd.DataFrame({"day": [ddmm(x) for x in cnt.index]})
pooled = []
for a, b, na, nb in zip(steps[:-1], steps[1:], names[:-1], names[1:]):
    step_rate[f"{na} → {nb}"] = (cnt[b] / cnt[a]).map("{:.3f}".format).to_numpy()
    rb, ri = B[b].notna().sum() / B[a].notna().sum(), I[b].notna().sum() / I[a].notna().sum()
    pooled.append({"step": f"{na} → {nb}", "15-17 Oct": f"{rb:.3f}", "18-19 Oct": f"{ri:.3f}", "change": f"{ri / rb - 1:+.1%}"})
show(step_rate)
print()
show(pd.DataFrame(pooled))
```

- **Observation:** the first step, Started Onboarding → Lead Form, falls the most: 0.315 to 0.249 (−20.8%).
- **Observation:** two later steps also fall: Strategy Session → Sales Call (−8.1%) and Sales Call → Closed Won (−5.8%). The other two barely move (−1.8% and −1.0%). Exercise 2 tests whether the larger of these later dips is a signal.

The first step is the largest, so it is examined first. Its own p-chart:

```python
pc = pchart(cnt["lead_form_submitted"], cnt["onboarding_start_date"], base_days)
pchart_table(pc)
for name, p_ in [("lead-form rate", pc), ("closed-won share", pc_won)]:
    r = p_.loc["2021-10-18"]
    print(f"18.10, {name}: one standard error = {(r['upper'] - r['centre']) / 3:.4f} = {(r['upper'] - r['centre']) / 3 / r['centre']:.1%} of {r['centre']:.3f}")
pchart_plot(pc, "share of starters submitting a lead form",
            "Onboarding → lead-form rate by onboarding day, p-chart (limits from 15-17 Oct)",
            "18.10 and 19.10 fall far below the lower limit;\n20.10 is back at the centre line", (1.0, 0.238), (0.22, 0.39))
```

- **Observation:** 18 Oct is 6.40 and 19 Oct is 5.81 standard errors below this step's baseline; the baseline days are within ±0.61, and 20 Oct is back at +0.04. This step's signal is stronger than the bottom line's (4.61 and 4.33) for two reasons: it is the largest single-step fall, and its limits are narrower relative to its rate (on 18.10, one standard error is 3.3% of 0.315, but 7.3% of 0.087).

**3b. Locate the segment.** The data has three ways to split starters: campaign, platform and browser. First, one cut at a time: each group's lead-form rate on baseline days vs. incident days.

```python
for dim, label in [("campaign_id", "campaign"), ("Platform", "platform"), ("browser", "browser")]:
    print(f"Started Onboarding → Lead Form rate, by {label} alone (15-17 Oct -> 18-19 Oct):")
    for k in sorted(d[dim].unique()):
        print(f"  {k} = {B[B[dim] == k]['lead_form'].mean():.3f} -> {I[I[dim] == k]['lead_form'].mean():.3f}")
    print()
```

- **Observation:** by campaign, only AAAA_123 falls (0.400 to 0.249); the other two hold. By platform, only Mobile falls (0.393 to 0.289); Desktop holds.
- **Observation:** by browser, Chrome and Safari both fall by about the same amount (0.318 to 0.250, and 0.308 to 0.249). A cut in which every group falls equally does not narrow the problem down, so browser drops out here (3c checks it once more). Read on its own, this cut would even look like a conversion drop everywhere: a single cut can make a mix shift look like a rate drop.

The two cuts that do narrow it down point at different things, AAAA_123 and Mobile. Crossing them shows whether it is one problem or two. Starters per day in each campaign × platform cell, and each cell's own lead-form rate:

```python
vol = d.pivot_table(index="cell", columns="ob_day", values="user_id", aggfunc="size")
vol.columns = [ddmm(c) for c in vol.columns]
show(vol, index=True)

cell_rate = pd.DataFrame({
    "cell": sorted(d["cell"].unique()),
    "rate 15-17 Oct": B.groupby("cell")["lead_form"].mean().map("{:.3f}".format).to_numpy(),
    "rate 18-19 Oct": I.groupby("cell")["lead_form"].mean().map("{:.3f}".format).to_numpy(),
    "starters 18-19 Oct": I.groupby("cell").size().to_numpy(),
})
rb, ri, ni = B.groupby("cell")["lead_form"].mean(), I.groupby("cell")["lead_form"].mean(), I.groupby("cell").size()
cell_rate["z (18-19 Oct vs. own baseline)"] = ((ri - rb) / np.sqrt(rb * (1 - rb) / ni)).map("{:+.2f}".format).to_numpy()
print()
show(cell_rate)
```

- **Observation:** AAAA_123 / Mobile fell from 264, 212 and 432 starters a day to 40 and 20 on 18–19 Oct, and came back to 212 on 20 Oct. On the same two days AAAA_123 / Desktop jumped from 308, 240 and 392 to 808 and 748. The other four cells' counts on 18–19 Oct are close to their 17 Oct counts, and their shares of the day stay inside normal limits (Part 5).
- **Observation:** inside every cell the lead-form rate is about the same on incident days as on baseline days: no cell's z is beyond ±3 (table above). AAAA_123 / Mobile even rises (0.562 to 0.633), but on only 60 starters.
- **Observation:** AAAA_123 / Mobile converts far better than any other cell: 0.562 on baseline days, against 0.245 for AAAA_123 / Desktop.

**3c. Mix vs. rate.** 3b found that one cell, AAAA_123 / Mobile, lost almost all its starters. Is that alone enough to explain the drop? Two "what ifs" test it:
- **mix change only:** each cell converts normally, but the cell mix is that day's.
- **rate change only:** the cell mix stays normal, but each cell converts as it did that day.

If "mix change only" matches "actual", the mix change explains the drop.

```python
def mix_vs_rate(col, title):
    """Per day and for 18-19 Oct pooled: actual rate vs. the two 'what if' rates."""
    base_share = B.groupby("cell").size() / len(B)
    base_rate = B.groupby("cell")[col].mean()
    groups = [(ddmm(day), g) for day, g in d.groupby("ob_day")] + [("18-19 Oct pooled", I)]
    rows = [{"day": name, "baseline": f"{B[col].mean():.4f}", "actual": f"{g[col].mean():.4f}",
             "mix change only": f"{((g.groupby('cell').size() / len(g)) * base_rate).sum():.4f}",
             "rate change only": f"{(base_share * g.groupby('cell')[col].mean()).sum():.4f}"} for name, g in groups]
    print(title + "\n")
    show(pd.DataFrame(rows))


mix_vs_rate("lead_form", "Started Onboarding → Lead Form rate (lead forms ÷ starters, the step where the fall starts):")
```

**Conclusion:** on 18.10 and 19.10, "mix change only" matches "actual". The first step's drop is explained by the mix change.

But the problem statement is about the closed-won rate, and two later steps fell too (3a). So the same test runs on the closed-won rate:

```python
mix_vs_rate("won", "Started Onboarding → Closed Won rate (closed-won ÷ starters, the Define measure):")
```

**Conclusion:** "mix change only" matches "actual" here too. The mix change explains the whole problem, not just the first step.

Why does losing one cell hurt the closed-won rate so much? Here is each cell's own closed-won rate:

```python
won_rate_b, won_rate_i = B.groupby("cell")["won"].mean(), I.groupby("cell")["won"].mean()
mix_b, mix_i = B.groupby("cell").size() / len(B), I.groupby("cell").size() / len(I)
t = pd.DataFrame({"cell": won_rate_b.index,
                  "users 15-17": mix_b.map("{:.1%}".format).to_numpy(), "users 18-19": mix_i.map("{:.1%}".format).to_numpy(),
                  "rate 15-17": won_rate_b.map("{:.4f}".format).to_numpy(), "rate 18-19": won_rate_i.map("{:.4f}".format).to_numpy(),
                  "mix change only": (mix_i * won_rate_b).map("{:.4f}".format).to_numpy(),
                  "rate change only": (mix_b * won_rate_i).map("{:.4f}".format).to_numpy()})
t.loc[len(t)] = ["total", "", "", "", "", f"{(mix_i * won_rate_b).sum():.4f}", f"{(mix_b * won_rate_i).sum():.4f}"]
print("Started Onboarding → Closed Won, by cell:\n")
show(t)
print("\nusers = % of all users from this cell, rate = closed-won rate (closed-won ÷ starters)")
print("mix change only = users 18-19 × rate 15-17")
print("rate change only = users 15-17 × rate 18-19")
print("total = the 6 cells added up")
```

0.0575 and 0.0866 are the "mix change only" and "rate change only" values in Window 2's 18-19 Oct pooled row.

**Observation:** the best cell, AAAA_123 / Mobile (closed-won rate 0.2048), fell from 22.4% to 1.6% of all users. AAAA_123 / Desktop (0.0638) took its place, rising from 23.2% to 41.4%. Every cell's own closed-won rate stayed about the same. So the total fell because users moved from a cell that closes 0.2048 of the time to one that closes 0.0638 of the time.

Could splitting the cells further, by browser, change the answer? Here is the same test with 12 cells instead of 6:

```python
inc_share = I.groupby("cell").size() / len(I)
cell3 = d["cell"] + " / " + d["browser"]
mix3 = ((I.groupby(cell3[I.index]).size() / len(I)) * B.groupby(cell3[B.index])["won"].mean()).sum()
print("Started Onboarding → Closed Won rate, 18-19 Oct pooled:")
print(f"  mix change only, 6 cells (campaign × platform) = {(inc_share * won_rate_b).sum():.4f}")
print(f"  mix change only, 12 cells (campaign × platform × browser) = {mix3:.4f}")
print(f"  actual = {w_inc:.4f}")
```

**Observation:** 0.0571 vs. 0.0575 is no real change. Browser adds nothing.

One last doubt: maybe the missing AAAA_123 mobile users didn't vanish, and the data just recorded them as desktop. If so, the desktop group would contain hidden mobile users. Mobile users submit lead forms far more often (0.562 vs. 0.245 for desktop, 3b table), so the desktop group would submit many more lead forms than normal. The window below checks.

```python
m_b = B[(B["campaign_id"] == "AAAA_123") & (B["Platform"] == "Mobile")]
m_i = I[(I["campaign_id"] == "AAAA_123") & (I["Platform"] == "Mobile")]
k_b = B[(B["campaign_id"] == "AAAA_123") & (B["Platform"] == "Desktop")]
k_i = I[(I["campaign_id"] == "AAAA_123") & (I["Platform"] == "Desktop")]
expected = len(m_b) / len(base_days) * len(inc_days)
missing = expected - len(m_i)
r_desk, r_mob = k_b["lead_form"].mean(), m_b["lead_form"].mean()
real_desk = len(k_i) - missing
lf_desk, lf_mob = round(real_desk * r_desk), round(missing * r_mob)
print("If the missing mobile users were recorded as desktop (AAAA_123, 18-19 Oct):")
print(f"  expected mobile users = {len(m_b):,} (15-17 Oct) ÷ 3 days × 2 days = {expected:,.0f}")
print(f"  actual mobile users = {len(m_i):,}")
print(f"  missing mobile users = {expected:,.0f} − {len(m_i):,} = {missing:,.0f}")
print()
print(f"  desktop users recorded = {len(k_i):,}")
print(f"    real desktop = {len(k_i):,} − {missing:,.0f} = {real_desk:,.0f}, × normal desktop rate {r_desk:.3f} = {lf_desk:,.0f} lead forms")
print(f"    hidden mobile = {missing:,.0f}, × normal mobile rate {r_mob:.3f} = {lf_mob:,.0f} lead forms")
print(f"  expected lead forms if the doubt were true = {lf_desk:,.0f} + {lf_mob:,.0f} = {lf_desk + lf_mob:,.0f} "
      f"(rate {lf_desk + lf_mob:,.0f} ÷ {len(k_i):,} = {(lf_desk + lf_mob) / len(k_i):.3f})")
print(f"  actual lead forms = {k_i['lead_form'].sum():,} (rate {k_i['lead_form'].mean():.3f})")
```

**Conclusion:** if mobile users were hiding in the desktop group, it would have submitted about 553 lead forms (rate 0.355). It submitted 364 (rate 0.234), close to the normal desktop rate. So the 545 mobile users really vanished, and the extra desktop users are real desktop users.

**Conclusion (mechanism):** the drop is a **traffic-mix shift, not a conversion failure**. For two days, AAAA_123 lost almost all its high-converting mobile traffic and gained low-converting desktop traffic. The total fell because the mix changed.

**3d. Mechanism vs. root cause.** The data proves the mechanism. It cannot say *why* AAAA_123's mobile traffic vanished, because ad delivery and landing-page data are not in it. The original investigation listed three candidate real-world causes, and all three stay **Hypotheses**:
- **H1a:** a campaign change (device targeting or budget) moved delivery from mobile to desktop, then was reverted. One action explains both the mobile collapse and the desktop surge.
- **H1b:** the mobile landing page broke, so mobile clicks never became onboarding starts, and the ad platform shifted delivery to desktop.
- **H1c:** a tracking break stopped mobile starts being recorded. This explains the mobile gap but not the desktop surge, so it is the weakest.

What would settle it: AAAA_123's ad-delivery report by device and day (impressions, clicks, spend), and the change log for the campaign and the mobile page. This is a correct and common place for an Analyze phase to end: **mechanism proven, root cause narrowed to named hypotheses, with the evidence that would decide between them.**

### Part 4 — Improve

For a two-day incident that reversed itself, "Improve" has two jobs. First, hand the fix to the owner of the cause (whoever runs campaign AAAA_123, to find and lock the change). Second, check the improvement **in the Define measure**.

```python
day20 = pc_won.loc["2021-10-20"]
share_mob = (vol.loc["AAAA_123 / Mobile"] / vol.sum()).rename("AAAA_123 / Mobile share")
print(f"20.10 closed-won share: {day20['rate']:.4f} (centre {day20['centre']:.4f}, z {day20['z']:+.2f}, inside limits)")
print("AAAA_123 / Mobile share of the day's starters: " + ", ".join(f"{k} = {v:.2%}" for k, v in share_mob.items()))

w_b18, w_b19 = cnt.loc["2021-10-18", "closed_won_date"], cnt.loc["2021-10-19", "closed_won_date"]
exp_won = w_base * len(I)
print("\nHow much did the incident cost the business? Closed-won deals lost on 18-19 Oct, compared with normal days:")
print(f"  starters on 18-19 Oct = {len(I):,}")
print(f"  normal closed-won rate (15-17 Oct) = {B['won'].sum():,} ÷ {len(B):,} = {w_base:.4f}")
print(f"  expected closed-won = {len(I):,} × {w_base:.4f} = {exp_won:,.0f}")
print(f"  actual closed-won = {w_b18} (18.10) + {w_b19} (19.10) = {I['won'].sum():,}")
print(f"  lost = {exp_won:,.0f} − {I['won'].sum():,} = {exp_won - I['won'].sum():,.0f}")
```

- **Observation:** on 20 Oct the closed-won share is 0.0881, at the centre line (+0.08), and AAAA_123 / Mobile is back to 22.22% of starters (from 1.11% on 19 Oct). Caveat: the recovery rests on one day, which may be partial (Part 0).
- **Observation:** the incident cost about 109 closed-won deals. This is an estimate: it assumes the two days would otherwise have looked like the baseline days.

**What a full Improve phase would still need:** the owner finds which hypothesis is true and fixes that cause. If H1a (a campaign edit), for example by requiring a review for device-targeting edits; if H1b (a broken page), by fixing it and adding page monitoring; if H1c (a tracking break), by fixing the tag and adding a check that the event still fires. The data here can't show that step. It can only confirm the recovery.

### Part 5 — Control: the phase the original investigation skipped

The original investigation *recommended* monitors ("alert on inputs, not just the blended funnel"), but did not build or test one. A Control phase does both.

The monitor watches the **cause**: each campaign × platform cell's share of daily starters, with p-chart limits from the same baseline days. A companion monitor watches each cell's own lead-form rate. Back-testing both on all six days shows whether they would have caught the incident, and how often they would have raised a false alarm.

```python
n_day = vol.sum()
alarms = []
for cell in vol.index:
    sh = pchart(vol.loc[cell], n_day, [ddmm(x) for x in base_days])
    g = d[d["cell"] == cell]
    rt = pchart(g.groupby("ob_day")["lead_form"].sum(), g.groupby("ob_day").size(), base_days)
    alarms.append({"cell": cell,
                   "share: days outside limits": ", ".join(sh.index[(sh["z"].abs() > 3)]) or "none",
                   "share: largest z": f"{sh['z'][sh['z'].abs().idxmax()]:+.2f}",
                   "own rate: days outside limits": ", ".join(ddmm(i) for i in rt.index[rt["z"].abs() > 3]) or "none",
                   "own rate: largest z": f"{rt['z'][rt['z'].abs().idxmax()]:+.2f}"})
show(pd.DataFrame(alarms))
```

```python
sh = pchart(vol.loc["AAAA_123 / Mobile"], n_day, [ddmm(x) for x in base_days])
fig, ax = plt.subplots(figsize=(7.5, 3.4))
x = np.arange(len(sh))
ax.step(x, sh["upper"], where="mid", color=GREY, lw=1, ls="--", label="control limits (±3 standard errors)")
ax.step(x, sh["lower"], where="mid", color=GREY, lw=1, ls="--")
ax.axhline(sh["centre"].iloc[0], color=GREY, lw=1, label=f"baseline share {sh['centre'].iloc[0]:.1%} (15-17 Oct)")
sig = sh["z"].abs() > 3
ax.plot(x, sh["rate"], color=BLUE, lw=1.5)
ax.scatter(x[~sig], sh["rate"][~sig], color=BLUE, zorder=3, label="day inside the limits")
ax.scatter(x[sig], sh["rate"][sig], color=ORANGE, zorder=3, s=50, label="day outside: alarm")
ax.set_xticks(x, sh.index)
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:.0%}"))
ax.set_ylabel("share of the day's starters")
ax.set_title("Control monitor back-test: AAAA_123 / Mobile share of daily starters, p-chart", fontsize=10, loc="left")
ax.text(3.5, 0.09, "the alarm fires on 18.10, the first incident day,\nand names the cell that changed", ha="center", fontsize=8.5, color="#333333")
ax.set_ylim(-0.01, 0.30)
ax.legend(fontsize=7.5, loc="lower left", frameon=False)
plt.tight_layout(); plt.show()
```

- **Observation:** the share monitor fires for exactly two cells, AAAA_123 / Mobile and AAAA_123 / Desktop, on exactly the two incident days (largest z −21.66 and +18.82: Mobile's share fell, Desktop's rose). It stays quiet for the other four cells on all six days: no false alarms in this back-test.
- **Observation:** no cell's own lead-form rate leaves its limits on any day (largest z −1.28). This is consistent with no conversion failure. One limit: AAAA_123 / Mobile had only 60 starters on the incident days, so its own-rate monitor could only have caught a very large change.
- **Caveat:** the back-test reuses the baseline days that set the limits, and covers only six days. Twelve monitors (6 cells × 2 measures) checked daily will raise an occasional false alarm over a longer run, by chance alone. The 3-standard-error limits keep that rare, not impossible.

**The control plan:**

| What is monitored | Measure | Trigger | Response | Owner | Demonstrated in |
|---|---|---|---|---|---|
| traffic mix | each campaign × platform cell's share of daily starters | outside p-chart limits (3 standard errors; initial baseline 15–17 Oct, re-set monthly) | check that campaign's delivery by device and its change log the same day | TBD (the data names no teams) | Part 5, share columns: fires on 18.10 and 19.10 for AAAA_123 only |
| conversion inside each cell | each cell's onboarding-to-lead-form rate | outside its p-chart limits | check that cell's landing page and onboarding flow | TBD | Part 5, own-rate columns: never fires |
| the headline | closed-won share of starters (the Define measure) | outside its p-chart limits | check the "traffic mix" and "conversion inside each cell" rows first: they show which cell changed, and whether it was the mix or the rate | TBD | Part 2, the closed-won p-chart: fires on 18.10 and 19.10 |
| later steps | each step's rate as a share of the previous step (e.g. Strategy Session → Sales Call) | outside its p-chart limits, daily and over the last 2 days pooled | check whether the cells feeding that step changed | TBD | Exercise 2, one step only (Strategy Session → Sales Call): no alarm daily, borderline with 2 days together (p = 0.011) |

The re-set period ("monthly") is a chosen example: limits from a short or stale baseline drift away from what normal looks like now.

### Part 6 — The tollgate scorecard

Each DMAIC deliverable, checked against what the original E01 reports contain.

| Phase | Deliverable | In the original E01? | Evidence / note |
|---|---|---|---|
| Define | problem statement (what, where, when, how big) | **Partly** | the Summary's first paragraph gives the size and dates, but already names the step (an Analyze result) |
| Define | scope, goal, SIPOC | **No** | not written; built in Part 1 |
| Measure | data checked against an independent total | **Yes** | recomputation matches the provided table (Part 0 repeats it) |
| Measure | data-quality issues surfaced first | **Yes** | out-of-order dates, the likely partial last day |
| Measure | baseline *with its normal variation* | **Partly** | a two-proportion z-test of pooled days, not a day-by-day control chart (built in Part 2) |
| Analyze | located step and segment | **Yes** | step table, campaign × platform split |
| Analyze | causes verified with data | **Yes** | mix-vs-rate split |
| Analyze | mechanism vs. root cause separated | **Yes** | three named hypotheses, plus the evidence that would decide |
| Improve | fix handed to the cause's owner, recovery checked in the Define measure | **Partly** | owner role named (marketing / ad operations), no person or team; recovery noted, but no check of the closed-won share (the Define measure) |
| Control | monitor built, back-tested, with trigger, response, owner | **No** | recommended in prose only; built and back-tested in Part 5 |

**Conclusion:** the original investigation was a strong Measure-and-Analyze piece. What DMAIC adds is the two ends: a written Define (a problem statement on the bottom line rather than the located step, plus scope and goal, so it is clear when the project is done), and a Control phase that is built and tested rather than recommended.

### Exercise 2 — Is the second, smaller dip a signal?

On 18–19 Oct a later step also dipped: Strategy Session → Sales Call (sales calls ÷ strategy sessions). Is that dip real, or normal noise? Two checks: each day on its own, then the two days together.

```python
ss = d[d["strategy_session_call_date"].notna()].copy()
ss["sales_call"] = ss["sales_call_date"].notna()
num, den = ss.groupby("ob_day")["sales_call"].sum(), ss.groupby("ob_day").size()
print("Check 1, each day on its own. Strategy Session → Sales Call rate, p-chart, limits from 15-17 Oct:\n")
pchart_table(pchart(num, den, base_days))
```

**Observation:** no single day is a signal (18.10 z = −2.58, 19.10 z = −2.34). Each day has only 265 or 239 strategy sessions, too few to show a small dip. Two days together have more:

```python
from statsmodels.stats.proportion import proportions_ztest
nb_, nb_all = num[base_days].sum(), den[base_days].sum()
ni_, ni_all = num[inc_days].sum(), den[inc_days].sum()
z2, p2val = proportions_ztest([ni_, nb_], [ni_all, nb_all])
print("Check 2, the two days together (two-proportion z-test). Strategy Session → Sales Call rate:\n")
print(f"  15-17 Oct = {nb_:,} ÷ {nb_all:,} = {nb_ / nb_all:.3f}")
print(f"  18-19 Oct = {ni_:,} ÷ {ni_all:,} = {ni_ / ni_all:.3f}")
print(f"  p = {p2val:.3f} (the chance of a gap this big if nothing had changed)")
print(f"  threshold = 0.05 ÷ 5 steps tested = {0.05 / 5:.2f} (stricter, because all 5 funnel steps were checked for a dip)")
```

**Observation:** p = 0.011, just above 0.01: borderline.

**Answer:** not clearly real. It is noise day by day and borderline over the two days. So it is not a finding, but something to watch: that is the control plan's "later steps" row.

## Key takeaways

- **DMAIC is a checklist of phases a complete investigation needs.** Define (what and how big, never why), Measure (trust the data, know what normal is), Analyze (locate, then verify causes), Improve (fix the cause, check in the Define measure), Control (keep it fixed).
- **A problem statement has no "why", no fix, and no located step in it.** The E01 one is on the bottom line: closed-won share 0.087 → 0.058 on 18–19 Oct, across 3,760 starters. Finding the step where the fall starts is Analyze's job.
- **A p-chart separates a signal from normal variation:** limits at the baseline rate ± 3 standard errors per day. The E01 incident days sat 4.61 and 4.33 standard errors below on the closed-won share (6.40 and 5.81 at the first step); the later Strategy → Sales dip did not cross the limits on any single day, and was only borderline with the two days together (p = 0.011, Exercise 2).
- **Mechanism is not root cause.** The data proved a traffic-mix shift ("mix change only" reproduces the lead-form rate, 0.2501 and 0.2474 vs. an actual 0.2478 and 0.2513, and the whole closed-won drop, 0.0575 vs. 0.0582). It could only narrow the real-world cause to three named hypotheses, plus the evidence that would decide.
- **Control means built and back-tested, not recommended.** A per-cell traffic-share monitor would have fired on the first incident day and named the cell, with no false alarms over the six days (an in-sample back-test: the limits came from three of those days).
- **The tollgate scorecard finds the gaps.** Here the original investigation was strong in Measure and Analyze; its gaps were a problem statement that already named the step, the written scope and goal, and the Control phase.
