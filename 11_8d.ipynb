## Lesson

### Where 8D comes from

Ford's 1987 manual *Team Oriented Problem Solving* (TOPS) set out the eight disciplines. In the 1990s a ninth step, D0, was added at the front for planning and emergency responses. 8D is now the standard way suppliers answer a customer complaint in manufacturing, and many companies' incident-review templates are 8D under another name.

### The nine steps

| step | name | the question it answers |
|---|---|---|
| D0 | Plan | is this an 8D problem, and does anything need doing in the next hours? |
| D1 | Team | who has the knowledge and the authority to solve it? |
| D2 | Describe the problem | what exactly is wrong, where, when, how big, and where is it **not**? |
| D3 | Interim containment | how do we protect the people affected **now**, before the cause is known? |
| D4 | Root cause **and escape point** | why did it happen, and why was it not caught sooner? |
| D5 | Permanent corrective action | what fix removes the root cause, and is there evidence it will work? |
| D6 | Implement and validate | did the fix work, measured on the problem itself? |
| D7 | Prevent recurrence | what must change so this, and problems like it elsewhere, do not happen again? |
| D8 | Recognise the team | close the work and credit the people who did it |

### What 8D adds

- **Containment first (D3).** DMAIC goes from Define straight to Measure and Analyze. 8D stops the damage first, with a temporary action, while the cause is still unknown.
- **Two root causes (D4).** One for why it happened. One for why it **escaped**: the escape point is "the closest point in the process where the root cause could have been found but was not" (Quality-One's definition).
- **Systemic prevention (D7).** Not only this process: every similar one, plus the documents that should have prevented it. These include procedures, the FMEA (Failure Mode and Effects Analysis, Module 10) and the **control plan** (the list of what is monitored, with its limit, owner and response).

### When to choose which

- **8D:** a specific incident, a clear start, someone affected now. Speed and containment matter.
- **DMAIC:** a chronic gap, such as a conversion rate that has always been too low. There is nothing to contain, and the work is mostly measuring and analysing.

## Worked example: the E01 funnel drop, run as an 8D

**Data choice: reuse, deliberately.** The same incident as Module 1: the E01 funnel data (E01 = an earlier take-home exercise by the user), where the closed-won rate fell on 2021-10-18 and 2021-10-19. Running the same incident through 8D gives a like-for-like comparison with Module 1's DMAIC version. Where Module 1 already proved something, this module reuses the result and says so.

**Where this was already done, unnamed:** the user's E01 investigation (`AI Analyst Workspace/Reports/E01_Summary.md`, a local file), and Module 1 of this playbook.

**Terms from Module 1:**
- **starter:** a user who started onboarding
- **closed-won deal:** a deal the sales team won
- **closed-won rate:** closed-won deals ÷ starters, per onboarding day. The **overall** closed-won rate counts all cells together. A cell's **own** closed-won rate counts one cell.
- **share:** used only for the traffic mix: a cell's starters ÷ all starters that day
- **cell:** one campaign × platform group, for example "AAAA_123 / Mobile"
- **baseline:** 15–17 Oct, pooled (all three days' deals ÷ all three days' starters). Chosen in Module 1 as the last normal days before the drop.
- **standard error:** how far chance alone typically moves a day's rate away from the baseline
- **z:** how many standard errors a day's rate sits from the baseline. Beyond 3 = a signal, as on a p-chart (a control chart for a proportion). |z| = z without its sign.
- **Module 1's three hypotheses** for the real-world cause:
  - **H1a:** a change to campaign AAAA_123 moved its ads from mobile to desktop, then was reverted
  - **H1b:** the mobile landing page of AAAA_123's ads broke
  - **H1c:** a tracking break stopped AAAA_123's mobile starts from being recorded

**Data caveat (from Module 1):** 20 Oct is the last day in the data and may be a partial day. 94 of the 103 closed-won deals from 15 Oct starters are dated 18 Oct, which looks like a dating quirk. Neither changes the shares and rates in this module, which count each starter on their onboarding day.

```python
import numpy as np
import pandas as pd
import common
from common import show

d = common.load_e01()
d["cell"] = d["campaign_id"] + " / " + d["Platform"]
d["won"] = d["closed_won_date"].notna()
BASE = ["2021-10-15", "2021-10-16", "2021-10-17"]
INC = ["2021-10-18", "2021-10-19"]
ddmm = lambda day: pd.Timestamp(day).strftime("%d.%m")
mz = lambda v: f"{v:+.2f}".replace("-", "−")

day = d.groupby("ob_day").agg(starters=("user_id", "size"), won=("won", "sum"))
pbar = day.loc[BASE, "won"].sum() / day.loc[BASE, "starters"].sum()
day["share"] = day["won"] / day["starters"]
day["z"] = (day["share"] - pbar) / np.sqrt(pbar * (1 - pbar) / day["starters"])
day["expected"] = day["starters"] * pbar
day["lost"] = day["expected"] - day["won"]

cells = d.groupby(["ob_day", "cell"]).size().unstack(fill_value=0)
tot = cells.sum(axis=1)
pb = cells.loc[BASE].sum() / tot[BASE].sum()
cshare = cells.div(tot, axis=0)
se = pd.DataFrame(np.sqrt(np.outer(1 / tot.values, (pb * (1 - pb)).values)), index=cshare.index, columns=cshare.columns)
cz = (cshare - pb) / se
print("Data loaded: E01 funnel, one row per starter")
print(f"  starters = {len(d):,}")
print(f"  onboarding days = {ddmm(day.index.min())} to {ddmm(day.index.max())}")
print(f"  baseline closed-won rate (15.10-17.10, pooled) = {pbar:.6f}")
w15 = d[(d["ob_day"] == "2021-10-15") & d["won"]]
print(f"  15.10 starters' closed-won deals dated 18.10 = {int((w15['closed_won_date'] == '2021-10-18').sum())} of {len(w15)}")
```

### D0 — Plan: is this an 8D problem?

8D fits a specific incident with a clear start. The first question is whether the drop is an incident or normal variation:

```python
print("Closed-won rate by onboarding day, against the baseline:\n")
show(pd.DataFrame({"day": [ddmm(x) for x in day.index], "starters": day["starters"].map("{:,}".format).to_numpy(),
                   "closed-won rate": day["share"].map("{:.4f}".format).to_numpy(), "z": day["z"].map(mz).to_numpy(),
                   "signal (|z| > 3)": np.where(day["z"].abs() > 3, "yes", "no")}))
```

**Observation:** 18 Oct, the first incident day, is a signal (z = −4.61), and so is 19 Oct (z = −4.33). The incident has a clear start.

**D0 decision:** this is an 8D problem: a sudden drop with a start date, hurting the sales pipeline now. (A drop that had always been there would be a DMAIC project.)

### D1 — Team

8D names a cross-functional team: the people who own each part of the problem. The data names no people, so the team is listed as proposed roles (names TBD = to be decided). Three of the five come from the three hypotheses, one per hypothesis:
- **the analyst:** owns the data and the measurements
- **the owner of campaign AAAA_123:** can check the campaign's device targeting and budget (H1a)
- **the owner of the mobile landing page:** can check whether it broke (H1b)
- **the owner of the tracking setup:** can check whether mobile starts stopped being recorded (H1c)
- **the sales lead:** the closed-won deals are theirs, and so is the impact

### D2 — Describe the problem: what it is, and what it is not

8D describes the problem with an is / is-not table (Module 4's technique): where the problem is, next to the closest place where it could be but is not. First, each cell's share of the starters, each cell's own closed-won rate, and the cost:

```python
inc_share = cells.loc[INC].sum() / tot[INC].sum()
t = pd.DataFrame({"cell": pb.index, "share 15.10-17.10": pb.map("{:.4f}".format).to_numpy(),
                  "share 18.10-19.10": inc_share.map("{:.4f}".format).to_numpy(),
                  "largest |z| on 18.10-19.10": cz.loc[INC].abs().max().map("{:.2f}".format).to_numpy()})
print("Each cell's share of the day's starters:\n")
show(t)
others = [c for c in cz.columns if not c.startswith("AAAA_123")]
print(f"\nlargest |z| among the other {len(others)} cells on 18.10-19.10 = {cz.loc[INC, others].abs().max().max():.2f}")
print()
rb = d[d["ob_day"].isin(BASE)].groupby("cell")["won"].agg(["mean", "size"])
ri = d[d["ob_day"].isin(INC)].groupby("cell")["won"].agg(["mean", "size"])
zr = (ri["mean"] - rb["mean"]) / np.sqrt(rb["mean"] * (1 - rb["mean"]) / ri["size"])
print("Each cell's own closed-won rate:\n")
show(pd.DataFrame({"cell": rb.index, "rate 15.10-17.10": rb["mean"].map("{:.4f}".format).to_numpy(),
                   "rate 18.10-19.10": ri["mean"].map("{:.4f}".format).to_numpy(), "starters 18.10-19.10": ri["size"].to_numpy(),
                   "z": zr.map(mz).to_numpy()}))
print(f"\nlargest |z| of a cell's own closed-won rate = {zr.abs().max():.2f}")
print()
print("Closed-won deals lost per incident day, compared with the baseline rate:")
for x in INC:
    r = day.loc[x]
    print(f"  {ddmm(x)}:")
    print(f"    expected = {int(r['starters']):,} × {pbar:.6f} = {r['expected']:.2f}")
    print(f"    actual = {int(r['won'])}")
    print(f"    lost = {r['expected']:.2f} − {int(r['won'])} = {r['lost']:.2f}")
print(f"  total lost = {day.loc[INC, 'lost'].sum():.2f}")
print(f"  average lost per incident day = {day.loc[INC, 'lost'].sum():.2f} ÷ {len(INC)} = {day.loc[INC, 'lost'].mean():.1f}")
```

| | the problem **is** | the problem **is not** |
|---|---|---|
| **what** | the overall closed-won rate (all cells together): 15.10–17.10 = 0.0874, 18.10 = 0.0580, 19.10 = 0.0585 | a fall in each cell's own closed-won rate: no cell's own rate moved beyond 3 standard errors (largest \|z\| = 0.49) |
| **where** | campaign AAAA_123: its recorded mobile starters' share fell from 0.2241 to 0.0160, its desktop starters' share rose from 0.2320 to 0.4138 | the other four cells: their shares stay within limits (largest \|z\| = 1.52, below 3) |
| **when** | onboarding days 18.10 and 19.10 | 15.10–17.10, and 20.10 (back at the baseline) |
| **how big** | about 109.49 closed-won deals lost, 54.7 a day on average (an estimate: it assumes each day would otherwise have looked like the baseline) | |

**Problem description:** for two onboarding days, 18–19 Oct, campaign AAAA_123's recorded mobile starters almost disappeared while its desktop starters rose, and the closed-won rate fell from 0.0874 to 0.0580 and 0.0585.

(Compared with Module 1's problem statement, which held only the bottom line, the 8D description also says where the problem is. In 8D, D2 includes locating it, so the team knows what to contain in D3.)

### D3 — Interim containment: protect the business now

The D2 tables already show the shape of the problem: one campaign's traffic moved, and no cell's own closed-won rate fell. Each incident day cost about 54.7 deals on average (an estimate). So the team acts before the cause is known (proposed actions, because they need no root cause):
1. **Tell the people who read the funnel report** that the drop is most likely a traffic-mix shift in one campaign, not a fall in any cell's own closed-won rate. This stops the wrong reaction, for example pressing the sales team about its close rate.
2. **Ask the owner of AAAA_123 to check, the same day, whether its ad delivery shifted from mobile to desktop**, and to undo any change found. If a campaign change (H1a) is the cause, this restores the traffic. If not, the check rules H1a out within hours, and the team moves on to the page (H1b) and the tracking (H1c).

While the containment is in place, the team also watches the six cells' shares every day (proposed): not containment itself, but the way to see whether it works.

**How to verify the containment (chosen):** the AAAA_123 / Mobile share returns to its normal range, |z| below 3.

**Hypothesis:** containment on the evening of 18 Oct could have saved most of 19 Oct's 51.82 lost deals. This assumes a same-day fix was possible.

### D4 — Root cause, and the escape point

**The mechanism (not yet the root cause).** Module 1 proved how the number moved: a traffic-mix shift. The check, repeated here: if every cell had kept its normal closed-won rate, but the cells' shares were those of 18–19 Oct, what would the closed-won rate have been?

```python
won_rate = d[d["ob_day"].isin(BASE)].groupby("cell")["won"].mean()
mix_only = (inc_share * won_rate).sum()
actual = day.loc[INC, "won"].sum() / day.loc[INC, "starters"].sum()
print("Closed-won rate on 18.10-19.10:\n")
print("mix change only = (share 18.10-19.10) × (rate 15.10-17.10), summed across all rows:\n")
show(pd.DataFrame({"cell": won_rate.index, "share 18.10-19.10": inc_share[won_rate.index].map("{:.4f}".format).to_numpy(),
                   "rate 15.10-17.10": won_rate.map("{:.4f}".format).to_numpy(),
                   "share × rate": (inc_share[won_rate.index] * won_rate).map("{:.4f}".format).to_numpy()}))
inc_won = [int(day.loc[x, "won"]) for x in INC]
inc_n = [int(day.loc[x, "starters"]) for x in INC]
print()
print(f"  mix change only = sum of the last column = {mix_only:.4f}")
print(f"  actual = closed-won ÷ starters = ({' + '.join(str(v) for v in inc_won)}) ÷ ({' + '.join(f'{v:,}' for v in inc_n)}) = {actual:.4f}")
print(f"  baseline = {pbar:.4f} (15.10-17.10, pooled)")
```

**Observation:** "mix change only" = 0.0575 predicts slightly more of a drop than happened (actual = 0.0582).

**Conclusion:** as in Module 1, the traffic-mix shift alone accounts for the drop, and leaves no room for a fall in any cell's own closed-won rate.

**Hypothesis:** the root cause, the real-world reason the traffic shifted, is one of H1a, H1b or H1c. The data cannot settle it, but it leans: AAAA_123's desktop starters rose, not only its mobile ones fell, and one campaign change (H1a) explains both, as Module 1 noted. AAAA_123's ad-delivery report by platform and day would settle it.

**The escape point.** Where could the problem have been caught, but was not? The closest point is the daily traffic mix itself. On 18 Oct, would a daily monitor have fired? Two kinds of monitor, checked on that day:

```python
first = "2021-10-18"
print(f"On {ddmm(first)}, the first incident day:")
print(f"  closed-won rate monitor: z = {mz(day.loc[first, 'z'])}")
for c in ["AAAA_123 / Mobile", "AAAA_123 / Desktop"]:
    print(f"  {c} cell-share monitor: z = {mz(cz.loc[first, c])}")
print(f"  largest |z| among the other {len(others)} cells' cell-share monitors = {cz.loc[first, others].abs().max():.2f}")
```

**Observation:** on 18 Oct, both kinds of monitor would have fired:
- the **closed-won rate monitor** (z = −4.61), which says only that the closed-won rate is lower, not where
- the **cell-share monitor**, for AAAA_123 / Mobile (z = −21.66) and AAAA_123 / Desktop (z = +18.82), which names the campaign and the platform

**Data caveat:** the cell shares add up to 100%, so the Mobile fall and the Desktop rise are one event, not two. And the z formula counts only chance, not the normal day-to-day swing of a campaign's traffic, so the size of these z values is overstated. Which cells moved, and in which direction, is not in doubt.

**Hypothesis:** no such monitor was running. The brief reported only "a significant drop in the company's main funnel", with no campaign or platform named, and the original investigation recommended monitors on inputs as a fix. Both point to no daily cell-share monitor, but the data cannot prove it.

**Escape point (if the Hypothesis holds):** no cell-share monitor: the traffic mix by campaign and platform was not watched daily. That is the closest point where the cause could have been seen, on its first day.

### D5 — Permanent corrective action

The fix depends on which hypothesis is true, so D5 starts (proposed) by finding out which of H1a, H1b and H1c it is, with AAAA_123's ad-delivery report by platform and day. Then (proposed, one per hypothesis, as in Module 1):
- **H1a, a campaign change:** undo the change, and require a review before anyone changes a campaign's device targeting.
- **H1b, a broken page:** fix the page, and add a check that it loads on mobile.
- **H1c, a tracking break:** fix the tracking, and add a check that the onboarding event still fires on mobile.

8D asks for evidence before rollout at two levels. The delivery report confirms the cause. Evidence that the chosen fix works would need a test of the fix itself, for example on one campaign first.

### D6 — Implement and validate

No permanent fix was put in place in this data, so D6 cannot be shown here. What the data can show is whether the problem stopped, on the two measures from D2: the closed-won rate, and the AAAA_123 / Mobile share.

```python
last = "2021-10-20"
print(f"On {ddmm(last)}:")
print(f"  closed-won rate = {day.loc[last, 'share']:.4f}")
print(f"  closed-won rate, baseline = {pbar:.4f}")
print(f"  closed-won rate, z = {mz(day.loc[last, 'z'])}")
print(f"  AAAA_123 / Mobile share = {cshare.loc[last, 'AAAA_123 / Mobile']:.4f}")
print(f"  AAAA_123 / Mobile share, baseline = {pb['AAAA_123 / Mobile']:.4f}")
print(f"  AAAA_123 / Mobile share, z = {mz(cz.loc[last, 'AAAA_123 / Mobile'])}")
print(f"  starters on {ddmm(last)} = {int(day.loc[last, 'starters']):,}")
```

**Observation:** on 20 Oct, both the closed-won rate and the AAAA_123 / Mobile share are back in their normal range (|z| below 3). That is consistent with the problem having stopped, not a validation: it is one day, with 954 starters, and the last day in the data. In a real 8D, D6 would run on the fixed process over several normal weeks.

### D7 — Prevent recurrence, beyond this incident

D7 asks what must change so that this problem, and problems like it elsewhere, cannot happen again unnoticed. Module 1's control plan already proposed the cell-share monitor, for every campaign × platform cell. D7 adds (proposed):
1. **Every new campaign, at launch:** its cells join the monitor from its first day, not after an incident.
2. **The documents:** the funnel's control plan (from Module 1) already holds the monitor. D7 adds "traffic-mix shift in one campaign" as a failure mode in the funnel's FMEA.
3. **The process behind the cause:** once D5 settles the hypothesis, apply its fix to every campaign or page of the same kind, for example the review step for device-targeting changes on all campaigns.

### D8 — Recognise the team

Close the 8D: record the description, the containment, both root causes and the fix, and credit the team. This step has no data. 8D's reasoning for it: people are more willing to report the next problem when the last one was handled fairly.

## DMAIC and 8D, side by side

### The same incident in both

Where each part of Module 1's DMAIC version sits in the 8D:

| 8D step | covered by Module 1 (DMAIC)? | what 8D added |
|---|---|---|
| D0 Plan | no | the decision that this is an incident, not a chronic gap |
| D1 Team | no | roles listed, three of them (one per hypothesis) |
| D2 Describe | partly: the Define statement, and Analyze's locating step | the problem located inside the description, as is / is-not |
| D3 Containment | **no** | actions for the same day, and the cost of each day waited |
| D4 Root cause | yes: the mix-vs-rate test and three hypotheses | **the escape point**: no cell-share monitor (a Hypothesis) |
| D5 Corrective action | yes: one fix per hypothesis | evidence for the cause, and for the fix, before rollout |
| D6 Validate | partly: the 20 Oct check that the problem stopped | validation of the fix itself, over several normal weeks (not possible here: no fix was built) |
| D7 Prevent | partly: the cell-share monitor for every cell | **every new campaign at launch**, the FMEA, and the process fix |
| D8 Recognise | no | closing and credit |

### How the two approaches differ

| | DMAIC | 8D |
|---|---|---|
| made for | a chronic gap in how a process performs | a specific incident that is hurting someone now |
| first move | define and measure the problem | decide it is an incident (D0), then contain it (D3) once the team (D1) and the description (D2) are in place |
| root causes | why it happened | why it happened, **and** why it escaped |
| statistics | at the centre (Measure, Analyze) | used where needed |
| ends with | a control plan | prevention across similar processes, and team recognition |
| who | a project team, led by one owner (often a Six Sigma Green or Black Belt: Six Sigma's training levels) | a cross-functional team, named in D1 |

**Conclusion:** for the E01 incident, 8D is the better fit: it had a clear start, and it was costing about 54.7 deals a day (an estimate).

**Hypothesis:** the fastest gain would have been containment, which DMAIC has no step for, if a same-day fix was possible. DMAIC's strengths, the careful baseline and the mix-vs-rate test, still did the work inside D0, D2 and D4.

## Exercise — The first evening

The curriculum's interview question, *"Your customers are seeing wrong numbers right now. What do you do first?"*, adapted to this incident: **the closed-won rate fell sharply today. What do you do first?**

It is the evening of 18 Oct. Only what is known by then can be used: starters up to 18 Oct, and closed-won deals dated up to 18 Oct.
1. Is there a signal, and does the data already point at a cell?
2. How many deals has the day cost so far?
3. Which 8D steps come first, and what is the first action taken?

**Data caveat:** with the full data, all 354 baseline deals turn out to be dated by 18 Oct. So the baseline that evening is the same 0.0874 used everywhere else.

```python
known = d[d["ob_day"] <= "2021-10-18"].copy()
known["won_by_18"] = known["closed_won_date"] <= "2021-10-18"
k18 = known[known["ob_day"] == "2021-10-18"]
kb = known[known["ob_day"].isin(BASE)]
base18 = kb["won_by_18"].mean()
share18 = k18["won_by_18"].mean()
z_won18 = (share18 - base18) / np.sqrt(base18 * (1 - base18) / len(k18))
print("Known by the evening of 18.10:")
print(f"  baseline closed-won deals known = {int(kb['won_by_18'].sum())}")
print(f"  closed-won rate on 18.10, known so far = deals known so far ÷ starters = {int(k18['won_by_18'].sum())} ÷ {len(k18):,} = {share18:.4f}")
print(f"  closed-won rate on 18.10, z = {mz(z_won18)}")
print(f"  expected deals on 18.10 = {len(k18):,} × {base18:.6f} = {len(k18) * base18:.2f}")
print(f"  deals known so far = {int(k18['won_by_18'].sum())}")
print(f"  gap so far = {len(k18) * base18:.2f} − {int(k18['won_by_18'].sum())} = {len(k18) * base18 - k18['won_by_18'].sum():.2f}")
c18 = known.groupby(["ob_day", "cell"]).size().unstack(fill_value=0)
s18 = c18.loc["2021-10-18"] / c18.loc["2021-10-18"].sum()
z18 = (s18 - pb) / np.sqrt(pb * (1 - pb) / c18.loc["2021-10-18"].sum())
print(f"  cell with the largest share fall = {z18.idxmin()}")
print(f"    z = {mz(z18.min())}")
print(f"  cell with the largest share rise = {z18.idxmax()}")
print(f"    z = {mz(z18.max())}")
```

**Answer:**
1. Yes. With the deals known that evening, the closed-won rate on 18 Oct is already a signal (z = −4.93). The cell shares, known as soon as the starters arrive, point at campaign AAAA_123: mobile fell (z = −21.66), desktop rose (z = +18.82).
2. A gap of 61.67 deals so far. It counts only the 110 deals known that evening (with the full data, 18 Oct's starters end with 114: 4 of their deals are dated after 18 Oct). So the final gap is smaller, 57.67.
